```python
from tqdm import tqdm
import numpy as np
import torch
import torchvision.transforms as transforms

import medmnist
from medmnist import DermaMNIST

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.model_selection import GridSearchCV

import torch.nn as nn
import torch.optim as optim
import time
```

```python
# =====================================
# PART 1: Data Exploration and Preprocessing
# =====================================

# Preprocessing - Taken from MedMNIST GitHub Repo
data_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[.5], std=[.5])
])

# Standard 28x28 version (MNIST-like) - Taken from MedMNIST GitHub Repo
train_dataset = DermaMNIST(split="train", transform=data_transform, download=True)
val_dataset   = DermaMNIST(split="val", transform=data_transform, download=True)
test_dataset  = DermaMNIST(split="test", transform=data_transform, download=True)

print("Training Dateset dimensions:\n")
print(train_dataset)
print("===================")
print("Val Dateset dimensions:\n")
print(val_dataset)
print("===================")
print("Test Dateset dimensions:\n")
print(test_dataset)
```

```python
# Save montage images
train_img = train_dataset.montage(length=20)
train_img.save("train-dataset-montage-20.png")
val_img = val_dataset.montage(length=20)
val_img.save("val-dataset-montage-20.png")
test_img = test_dataset.montage(length=20)
test_img.save("test-dataset-montage-20.png")
```

```python
class_names = list(train_dataset.info['label'].values())
num_classes = len(class_names)
num_channels = train_dataset.info['n_channels']

def get_label_counts(dataset):
    counts = {}
    for i in range(len(dataset)):
        label = dataset[i][1].item()
        if label in counts:
            counts[label] += 1
        else:
            counts[label] = 1
    return counts

train_counts = get_label_counts(train_dataset)
val_counts = get_label_counts(val_dataset)
test_counts = get_label_counts(test_dataset)

print("Class Distribution for the 3 datasets (class name, number of samples): \n")
print("===================")
print("1. Train Dataset\n")
for i, name in enumerate(class_names):
    print(f"{name}, {train_counts[i]}")

print("===================")    
print("2. Val Dataset\n")
for i, name in enumerate(class_names):
    print(f"{name}, {train_counts[i]}")

print("===================")
print("3. Test Dataset\n")
for i, name in enumerate(class_names):
    print(f"{name}, {train_counts[i]}")
```

```python
# 2. Train a multi-class Logistic Regression model

# Initialize model
model = LogisticRegression(solver='lbfgs', max_iter=2000)
print("Initialized model.")

# Train the model
model.fit(X_train, y_train)
print("Trained model.")

# Evaluate on Validation or Test set
# y_pred = model.predict(X_test)
# print(f"Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
# print(classification_report(y_test, y_pred))

# 2. Define the search space (powers of 10 are a standard starting point)
param_grid = {'C': [0.01, 0.1, 1, 10]}

# 3. Initialize the grid search
# 'cv=5' means 5-fold cross-validation
grid_search = GridSearchCV(model, param_grid, cv=5, scoring='accuracy', verbose=2)

# 4. Run the search
print("Running grid search... ")
grid_search.fit(X_train, y_train)
print("Finished grid search.")

# 5. Get the results
print(f"Best C value: {grid_search.best_params_['C']}")
print(f"Best Cross-Validation Accuracy: {grid_search.best_score_:.4f}")
```

```python
# =====================================
# PART 3: Neural Network
# =====================================

import matplotlib.pyplot as plt

#Define Architecture
class DermaMNIST_NN(nn.Module):
    def __init__(self, input_size, hidden_size, num_classes):
        super(DermaMNIST_NN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)   # W^(1), b^(1)
        self.relu = nn.ReLU()                           # f(z^(1))
        self.fc2 = nn.Linear(hidden_size, num_classes)  # W^(2), b^(2)
        # Note: softmax is handled internally by CrossEntropyLoss

    def forward(self, x):
        z1 = self.fc1(x)           # z^(1) = W^(1)x + b^(1) = first linear layer
        a1 = self.relu(z1)         # a^(1) = f(z^(1)) = activation function (ReLU)
        z2 = self.fc2(a1)          # z^(2) = W^(2)a^(1) + b^(2) = second linear layer
        return z2                  # softmax applied in loss function

INPUT_SIZE   = 2352   # 3 * 28 * 28 (flattened image)
HIDDEN_SIZE  = 128    # neurons in hidden layer
NUM_CLASSES  = 7      # DermaMNIST classes
LEARNING_RATE = 0.001
NUM_EPOCHS   = 20

# Initialize model
model = DermaMNIST_NN(INPUT_SIZE, HIDDEN_SIZE, NUM_CLASSES)
print(model)
print(f"\nTrainable parameters: {sum(p.numel() for p in model.parameters()):,}")

# Loss Function and optimizers
criterion = nn.CrossEntropyLoss()    # L = -Σ y_i log(ŷ_i) = Cross Entropy Loss (includes softmax internally, used to output probabilities over 7 classes)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)    # w ← w - η * ∂L/∂w = SGD / Adam gradient update

# Training
train_losses, val_losses     = [], []
train_accuracies, val_accuracies = [], []

start_time = time.time()
for epoch in range(NUM_EPOCHS):
    # Training process
    model.train()
    correct, total, running_loss = 0, 0, 0.0

    for images, labels in train_loader:
        images = images.view(images.size(0), -1)   # flatten: (B, 2352)
        labels = labels.squeeze().long()

        optimizer.zero_grad()

        # Forward pass
        outputs = model(images)     
        
        # Loss function
        loss = criterion(outputs, labels)
        
        # Backpropagation
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        correct += (predicted == labels).sum().item()
        total   += labels.size(0)

    train_loss = running_loss / len(train_loader)
    train_acc  = correct / total
    train_losses.append(train_loss)
    train_accuracies.append(train_acc)

    # Validation
    model.eval()
    correct, total, running_loss = 0, 0, 0.0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.view(images.size(0), -1)
            labels = labels.squeeze().long()
            outputs = model(images)
            loss = criterion(outputs, labels)
            running_loss += loss.item()
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == labels).sum().item()
            total   += labels.size(0)

    val_loss = running_loss / len(val_loader)
    val_acc  = correct / total
    val_losses.append(val_loss)
    val_accuracies.append(val_acc)

    print(f"Epoch [{epoch+1:02d}/{NUM_EPOCHS}]  "
          f"Train Loss: {train_loss:.4f}  Train Acc: {train_acc:.4f}  "
          f"Val Loss: {val_loss:.4f}  Val Acc: {val_acc:.4f}")

total_time = time.time() - start_time
print(f"\nTotal training time: {total_time:.2f}s  ({total_time/60:.2f} mins)")

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4))

ax1.plot(train_losses, label="Train")
ax1.plot(val_losses,   label="Validation")
ax1.set_title("Loss per Epoch")
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Loss")
ax1.legend()

ax2.plot(train_accuracies, label="Train")
ax2.plot(val_accuracies,   label="Validation")
ax2.set_title("Accuracy per Epoch")
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Accuracy")
ax2.legend()

plt.tight_layout()
plt.savefig("nn_training_curves.png", dpi=150)
plt.show()

# Evaluate metrics
model.eval()
all_preds, all_labels = [], []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.view(images.size(0), -1)
        labels = labels.squeeze().long()
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        all_preds.extend(predicted.numpy())
        all_labels.extend(labels.numpy())

test_acc = accuracy_score(all_labels, all_preds)
print(f"\nTest Accuracy: {test_acc:.4f}")
print(classification_report(all_labels, all_preds, target_names=class_names))
```
